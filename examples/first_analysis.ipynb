# Первый анализ продаж

Синтетические данные: шесть строк. Цель — рассчитать общую выручку и выручку по товарам.

```python
from pathlib import Path
import pandas as pd
data = pd.read_csv(Path('sales.csv'))
data
```

```python
print('Размер таблицы:', data.shape)
print(data.dtypes)
print('Пропуски:', int(data.isna().sum().sum()))
assert len(data) == 6
assert not data.isna().any().any()
```

## Расчёт

Выручка строки равна количеству единиц, умноженному на цену одной единицы. Затем суммируем строки каждого товара.

```python
data['revenue'] = data['quantity'] * data['unit_price']
by_product = data.groupby('product', as_index=False)['revenue'].sum()
by_product = by_product.sort_values('revenue', ascending=False).reset_index(drop=True)
by_product
```

```python
total = int(data['revenue'].sum())
print('Общая выручка:', total, 'руб.')
assert total == 4100
assert by_product.loc[0, 'product'] == 'Книга'
assert by_product.loc[0, 'revenue'] == 2000
```

## Вывод

Общая выручка по шести продажам составила 4 100 рублей. Наибольшая выручка у товара «Книга»: 2 000 рублей. Маленький синтетический набор не позволяет делать выводы о реальном спросе.
